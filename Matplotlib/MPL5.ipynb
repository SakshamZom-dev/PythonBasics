```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20, 35, 30, 50, 45]

plt.figure(figsize=(8, 4))
plt.plot(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20, 35, 30, 50, 45]

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["A", "B", "C", "D", "E"]
scores = [45, 60, 75, 85, 95]

plt.plot(students, scores, marker="o")

plt.ylim(0, 100)            # Set y-axis limits
plt.xlim(0, 4)              # Set x-axis limits

plt.title("Student Scores")
plt.xlabel("Students")
plt.ylabel("Scores")

plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["Mathematics", "Programming", "Database", "Statistics"]
marks = [75, 90, 65, 80]

plt.bar(subjects, marks)

plt.xticks(rotation=45)                 # Rotate x-axis labels for better readability
plt.yticks([0, 20, 40, 60, 80, 100])    # Set y-axis ticks

plt.title("Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5]
scores = [35, 45, 60, 70, 85]

plt.scatter(
    hours,
    scores,
    color="purple",
    marker="^",
    s=100,
    alpha=0.5   # Transparency
)

plt.title("Study Hours vs Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Scores")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20, 35, 30, 50, 45]

plt.plot(months, sales, color="green", marker="s", linestyle="--", linewidth=2)

plt.title("Monthly Sales Data")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
```
