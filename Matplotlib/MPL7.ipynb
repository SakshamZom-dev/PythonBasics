```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20, 35, 30, 50, 45]

fig, ax = plt.subplots(figsize=(8, 4))  # Create a figure and axis with specified size

ax.plot(months, sales, marker="o")
ax.set_title("Monthly Sales")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")

plt.show()
```

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
temperatures = [28, 30, 29, 32, 31]

fig, ax = plt.subplots(figsize = (10, 5))

ax.plot(days, temperatures, marker = "s", color = "green")

ax.grid(True)

ax.set_title("Weekly Temperatures")
ax.set_xlabel("Days")
ax.set_ylabel("Temperatures")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [20, 35, 30, 50]
profit = [5, 10, 8, 15]

fig, axes = plt.subplots(1, 2, figsize=(10, 4)) # Create subplots, 1 row, 2 columns, size 10x4

axes[0].plot(months, sales, marker="o")         # Sales plot
axes[0].set_title("Monthly Sales")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Sales")

axes[1].bar(months, profit, color="green")      # Profit bar chart
axes[1].set_title("Monthly Profit")
axes[1].set_xlabel("Month")
axes[1].set_ylabel("Profit")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales_2025 = [20, 30, 25, 40, 50]
sales_2026 = [25, 35, 40, 45, 60]

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

axes[0].plot(months, sales_2025, marker="o")
axes[0].set_title("Monthly Sales (2025)")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Sales")

axes[1].plot(months, sales_2026, marker="o", color="orange")
axes[1].set_title("Monthly Sales (2026)")
axes[1].set_xlabel("Month")
axes[1].set_ylabel("Sales")

plt.tight_layout()
plt.show()
```
