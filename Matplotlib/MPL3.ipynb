```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6]
marks = [40, 50, 55, 65, 75, 85]

plt.scatter(study_hours, marks)

plt.title("Study Hours vs Exam Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6]
marks = [40, 50, 55, 65, 75, 85]

plt.scatter(
    study_hours,
    marks,
    color="purple",
    s=100
)
plt.title("Study Hours vs Exam Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6]
marks = [40, 50, 55, 65, 75, 85]

plt.scatter(
    study_hours,
    marks,
    color="green",
    marker="^",
    s=100
)
plt.title("Study Hours vs Exam Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```
