```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
sales = [1000, 1500, 1200, 2000, 2500]

plt.plot(days, sales)

plt.title("Daily Sales")
plt.xlabel("Day")
plt.ylabel("Sales (₹)")

plt.show()
```

```python
# Customizing color and style

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [1200, 1800, 1500, 2400, 3000]

plt.plot(months, sales, color="green", marker="o", linestyle="--")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [1200, 1800, 1500, 2400, 3000]

plt.plot(
    months,
    sales,
    color = "red",
    marker =  "s",
    linestyle = "--"
)

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

laptop_sales = [1200, 1500, 1400, 2000, 2400]
phone_sales = [1800, 1600, 2200, 1900, 2600]

plt.plot(
    months,
    laptop_sales,
    color="blue",
    marker="o"
)

plt.plot(
    months,
    phone_sales,
    color="orange",
    marker="s"
)

plt.title("Monthly Sales Comparison")
plt.xlabel("Months")
plt.ylabel("Sales")


plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

laptop_sales = [1200, 1500, 1400, 2000, 2400]
phone_sales = [1800, 1600, 2200, 1900, 2600]

plt.plot(
    months,
    laptop_sales,
    color="blue",
    marker="o",
    label="Laptops"
)

plt.plot(
    months,
    phone_sales,
    color="orange",
    marker="s",
    label="Phones"
)

plt.title("Monthly Product Sales")
plt.xlabel("Month")
plt.ylabel("Units Sold")

plt.legend()
plt.show()
```
