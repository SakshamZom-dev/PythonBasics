```python
import matplotlib.pyplot as plt

categories = ["Food", "Travel", "Shopping", "Other"]
expenses = [4000, 2000, 3000, 1000]

plt.pie(expenses, labels=categories)

plt.title("Monthly Expenses")
plt.show()
```

```python
import matplotlib.pyplot as plt

categories = ["Food", "Travel", "Shopping", "Other"]
expenses = [4000, 2000, 3000, 1000]

plt.pie(
    expenses,
    labels=categories,
    autopct="%1.1f%%"
)
plt.title("Monthly Expenses")
plt.show()
```

```python
import matplotlib.pyplot as plt

languages = ["Python", "Java", "C++", "JavaScript"]
students = [40, 25, 15, 20]

plt.pie(
    students,
    labels=languages,
    autopct="%1.1f%%"
)

plt.title("Programming Language Preferences")

plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 65, 70, 75, 80, 95]

plt.boxplot(marks)

plt.title("Distribution of Exam Marks")
plt.ylabel("Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

scores = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70, 95]

plt.boxplot(scores)

plt.title("Exam Score Distribution")
plt.ylabel("Scores")

plt.show()
```

```python
import matplotlib.pyplot as plt

class_a = [45, 50, 55, 60, 65, 70, 75]
class_b = [55, 60, 65, 70, 75, 80, 85]

plt.boxplot([class_a, class_b], tick_labels=["Class A", "Class B"])

plt.title("Comparison of Exam Scores")
plt.ylabel("Scores")

plt.show()
```
