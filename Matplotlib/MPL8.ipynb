```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales_2025 = [20, 30, 25, 40, 50]
sales_2026 = [25, 35, 40, 45, 60]

plt.plot(months, sales_2025, marker="o", label="2025")
plt.plot(months, sales_2026, marker="s", label="2026")

plt.title("Sales Comparison: 2025 vs 2026")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

months = ["Jan", "Feb", "Mar", "Apr"]
product_a = [20, 35, 30, 40]
product_b = [25, 30, 35, 45]

x = np.arange(len(months))
width = 0.35

plt.bar(x - width/2, product_a, width, label="Product A", color="violet")
plt.bar(x + width/2, product_b, width, label="Product B", color="purple")

plt.xticks(x, months)
plt.title("Product Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()

plt.show()
```
