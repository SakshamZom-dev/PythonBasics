```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20, 35, 30, 50, 45]

plt.plot(months, sales, marker="o")

plt.axhline(y=40, color="red", linestyle="--")

plt.title("Monthly Sales vs Target")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20, 35, 30, 50, 45]

plt.plot(months, sales, marker="o")

plt.annotate(
    "Highest sales",
    xy=("Apr", 50),
    xytext=("Feb", 55),
    arrowprops=dict(arrowstyle="->")
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
import matplotlib.pyplot as plt

products = ["A", "B", "C", "D", "E"]
sales = [30, 45, 60, 40, 75]

plt.plot(products, sales, marker="o")

plt.axhline(y=50, color="red", linestyle="--")

plt.annotate(
    "Highest Sales",
    xy=("E", 75),
    xytext=("D", 80),
    arrowprops=dict(arrowstyle="->")
)

plt.title("Monthly Sales vs Target")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
```
